## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Bagaimana perkembangan total pendapatan dan jumlah pesanan setiap bulan selama periode operasional Olist?
- **Pertanyaan 2:** Kategori produk apa yang memberikan kontribusi pendapatan terbesar selama periode pengamatan?
- **Pertanyaan 3:** Bagaimana karakteristik pelanggan berdasarkan Recency, Frequency, dan Monetary (RFM), serta bagaimana distribusi masing-masing segmen pelanggan?
- **Pertanyaan 4:** Bagaimana hubungan ketepatan waktu pengiriman dengan skor ulasan pelanggan selama periode pengamatan?

## Exploratory Data Analysis (EDA)

### Explore ...


Pada tahap Exploratory Data Analysis (EDA), dilakukan
eksplorasi terhadap dataset yang telah melalui proses
cleaning untuk memperoleh informasi dan menemukan pola
yang dapat menjawab pertanyaan bisnis.

Analisis dilakukan menggunakan teknik statistik
deskriptif, grouping, agregasi, dan analisis lanjutan
berupa RFM Analysis.

Hasil eksplorasi akan digunakan untuk memahami
perkembangan transaksi, kontribusi kategori produk,
karakteristik pelanggan, serta hubungan ketepatan

```python

# Menggabungkan data pesanan dengan detail produk
sales_df = pd.merge(
    orders_clean,
    order_items_clean,
    on='order_id',
    how='inner'
)

# Menggabungkan dengan data pelanggan
sales_df = pd.merge(
    sales_df,
    customers_clean[['customer_id', 'customer_unique_id']],
    on='customer_id',
    how='left'
)

# Menggabungkan dengan data produk
sales_df = pd.merge(
    sales_df,
    products_clean[['product_id', 'product_category_name']],
    on='product_id',
    how='left'
)

# Menggabungkan dengan terjemahan kategori produk
sales_df = pd.merge(
    sales_df,
    category_translation_clean,
    on='product_category_name',
    how='left'
)

# Memilih pesanan yang sudah selesai
sales_df = sales_df[
    sales_df['order_status'] == 'delivered'
].copy()

# Mengubah kategori yang belum memiliki terjemahan
sales_df['product_category_name_english'] = (
    sales_df['product_category_name_english']
    .fillna('unknown')
)

print("Data analisis berhasil disiapkan!")
print("Jumlah baris:", sales_df.shape[0])
print("Jumlah kolom:", sales_df.shape[1])

sales_df.head()
```


### Analisis 1: Tren Nilai Transaksi dan Jumlah Pesanan Bulanan

Analisis ini bertujuan untuk mengetahui perkembangan
nilai transaksi dan jumlah pesanan setiap bulan.

Pengelompokan dilakukan berdasarkan tanggal pembelian
(order_purchase_timestamp). Nilai transaksi dihitung
berdasarkan total harga produk (price), sedangkan
jumlah pesanan dihitung berdasarkan jumlah order_id
yang unik.

Analisis ini menggunakan pesanan dengan status
delivered agar perhitungan berfokus pada transaksi
yang telah selesai dikirim.


```python
# Mengelompokkan transaksi berdasarkan bulan
monthly_sales = sales_df.groupby(
    sales_df['order_purchase_timestamp'].dt.to_period('M')
).agg(
    total_revenue=('price', 'sum'),
    total_orders=('order_id', 'nunique')
).reset_index()

# Mengubah periode menjadi format tanggal
monthly_sales['order_purchase_timestamp'] = (
    monthly_sales['order_purchase_timestamp'].astype(str)
)

# Menampilkan hasil analisis
monthly_sales.head(10)
```

```python

# Bulan dengan nilai transaksi tertinggi
highest_revenue = monthly_sales.loc[
    monthly_sales['total_revenue'].idxmax()
]

# Bulan dengan nilai transaksi terendah
lowest_revenue = monthly_sales.loc[
    monthly_sales['total_revenue'].idxmin()
]

print("Bulan dengan nilai transaksi tertinggi:")
print(highest_revenue)

print("\nBulan dengan nilai transaksi terendah:")
print(lowest_revenue)
```


### Analisis 2: Kontribusi Nilai Transaksi Berdasarkan Kategori Produk

Analisis ini dilakukan untuk mengetahui kategori produk
yang memberikan kontribusi nilai transaksi terbesar.

Pengelompokan dilakukan berdasarkan kategori produk
dengan menghitung total harga produk dari seluruh
pesanan yang telah berstatus delivered.

Hasil analisis digunakan untuk mengidentifikasi
kategori produk dengan kontribusi transaksi terbesar
dan terkecil.


```python
# Mengelompokkan transaksi berdasarkan kategori produk
category_sales = sales_df.groupby(
    'product_category_name_english'
).agg(
    total_revenue=('price', 'sum'),
    total_products=('product_id', 'count'),
    total_orders=('order_id', 'nunique')
).reset_index()

# Mengurutkan kategori berdasarkan nilai transaksi
category_sales = category_sales.sort_values(
    by='total_revenue',
    ascending=False
)

# Menghitung persentase kontribusi
category_sales['revenue_percentage'] = (
    category_sales['total_revenue'] /
    category_sales['total_revenue'].sum()
) * 100

# Menampilkan 10 kategori teratas
category_sales.head(10)
```

```python
# Kategori dengan nilai transaksi tertinggi
highest_category = category_sales.iloc[0]

# Kategori dengan nilai transaksi terendah
lowest_category = category_sales.iloc[-1]

print("Kategori dengan nilai transaksi tertinggi:")
print(highest_category)

print("\nKategori dengan nilai transaksi terendah:")
print(lowest_category)
```


### Analisis 3: Segmentasi Pelanggan Menggunakan RFM Analysis

RFM Analysis digunakan untuk mengelompokkan pelanggan
berdasarkan tiga indikator, yaitu:

1. Recency: jumlah hari sejak transaksi terakhir
   pelanggan.
2. Frequency: jumlah pesanan yang dilakukan pelanggan.
3. Monetary: total nilai pembelian pelanggan.

Analisis menggunakan customer_unique_id agar
pelanggan yang sama dapat dikenali meskipun memiliki
lebih dari satu customer_id.

Skor RFM akan dibagi menjadi lima kelompok berdasarkan
kuintil. Hasilnya digunakan untuk mengidentifikasi
karakteristik dan segmentasi pelanggan.


```python
# Menentukan tanggal acuan analisis
reference_date = (
    sales_df['order_purchase_timestamp'].max()
    + pd.Timedelta(days=1)
)

# Menghitung RFM setiap pelanggan
rfm_df = sales_df.groupby('customer_unique_id').agg(
    recency=('order_purchase_timestamp',
             lambda x: (reference_date - x.max()).days),
    frequency=('order_id', 'nunique'),
    monetary=('price', 'sum')
).reset_index()

# Menampilkan hasil RFM
rfm_df.head()
```

```python
# Membuat skor berdasarkan lima kelompok
# Recency semakin kecil semakin baik
rfm_df['R_Score'] = pd.qcut(
    rfm_df['recency'].rank(method='first'),
    q=5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

# Frequency semakin besar semakin baik
rfm_df['F_Score'] = pd.qcut(
    rfm_df['frequency'].rank(method='first'),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Monetary semakin besar semakin baik
rfm_df['M_Score'] = pd.qcut(
    rfm_df['monetary'].rank(method='first'),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Menghitung total skor RFM
rfm_df['RFM_Score'] = (
    rfm_df['R_Score'].astype(str) +
    rfm_df['F_Score'].astype(str) +
    rfm_df['M_Score'].astype(str)
)

rfm_df.head()
```

```python
def segment_customer(row):
    r = row['R_Score']
    f = row['F_Score']
    m = row['M_Score']

    if r >= 4 and f >= 4 and m >= 4:
        return 'Best Customers'
    elif r >= 3 and f >= 4:
        return 'Loyal Customers'
    elif r >= 4 and f >= 2:
        return 'Potential Loyalists'
    elif r <= 2 and f >= 3:
        return 'At Risk Customers'
    elif r <= 2 and f <= 2:
        return 'Lost Customers'
    else:
        return 'Regular Customers'

rfm_df['customer_segment'] = rfm_df.apply(
    segment_customer,
    axis=1
)

# Menghitung jumlah pelanggan setiap segmen
segment_summary = rfm_df.groupby(
    'customer_segment'
).agg(
    total_customers=('customer_unique_id', 'count'),
    average_recency=('recency', 'mean'),
    average_frequency=('frequency', 'mean'),
    average_monetary=('monetary', 'mean')
).reset_index()

segment_summary = segment_summary.sort_values(
    by='total_customers',
    ascending=False
)

segment_summary
```


### Analisis 4: Hubungan Ketepatan Pengiriman dengan Skor Ulasan

Analisis ini bertujuan untuk membandingkan rata-rata
skor ulasan pelanggan berdasarkan ketepatan waktu
pengiriman pesanan.

Pesanan dikategorikan menjadi dua kelompok, yaitu
tepat waktu dan terlambat berdasarkan perbandingan
tanggal pesanan diterima dengan estimasi tanggal
pengiriman.

Selanjutnya, kedua kelompok dibandingkan berdasarkan
rata-rata review_score untuk mengetahui perbedaan
penilaian pelanggan.


```python
# Mengambil data pesanan yang sudah terkirim
delivery_df = orders_clean[
    orders_clean['order_status'] == 'delivered'
][[
    'order_id',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]].copy()

# Menghitung rata-rata skor ulasan per pesanan
review_per_order = order_reviews_clean.groupby(
    'order_id'
).agg(
    review_score=('review_score', 'mean')
).reset_index()

# Menggabungkan data pengiriman dengan ulasan
delivery_review_df = pd.merge(
    delivery_df,
    review_per_order,
    on='order_id',
    how='inner'
)

# Menghapus data yang tidak memiliki tanggal atau skor ulasan
delivery_review_df = delivery_review_df.dropna(
    subset=[
        'order_delivered_customer_date',
        'order_estimated_delivery_date',
        'review_score'
    ]
)

print("Jumlah data yang dapat dianalisis:",
      len(delivery_review_df))

delivery_review_df.head()
```

#### 1.1 Menentukan Batas Kelompok Pelanggan

Pengelompokan pelanggan dilakukan berdasarkan nilai Monetary menggunakan teknik binning. Batas kategori ditentukan berdasarkan distribusi data sehingga pelanggan dapat dikelompokkan menjadi Low Value Customer, Medium Value Customer, dan High Value Customer.

```python
# Melihat statistik deskriptif nilai Monetary
rfm_df['monetary'].describe()
```

```python
# Menghitung batas kelompok berdasarkan persentil
lower_limit = rfm_df['monetary'].quantile(0.33)
upper_limit = rfm_df['monetary'].quantile(0.67)

print("Batas Low Value dan Medium Value:", lower_limit)
print("Batas Medium Value dan High Value:", upper_limit)
```

```python
# Membuat kolom customer_value_segment
rfm_df['customer_value_segment'] = pd.cut(
    rfm_df['monetary'],
    bins=[-float('inf'), lower_limit, upper_limit, float('inf')],
    labels=['Low Value Customer', 'Medium Value Customer', 'High Value Customer'],
    include_lowest=True
)

# Melihat hasil pengelompokan
rfm_df[['customer_unique_id', 'monetary', 'customer_value_segment']].head(10)
```

```python
# Menghitung jumlah pelanggan berdasarkan kategori
customer_value_summary = (
    rfm_df['customer_value_segment']
    .value_counts()
    .reindex(['Low Value Customer', 'Medium Value Customer', 'High Value Customer'])
    .reset_index()
)

customer_value_summary.columns = ['customer_value_segment', 'total_customers']

# Menghitung persentase pelanggan
customer_value_summary['percentage'] = (
    customer_value_summary['total_customers'] /
    customer_value_summary['total_customers'].sum() * 100
)

customer_value_summary
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=customer_value_summary,
    x='customer_value_segment',
    y='total_customers'
)

plt.title('Customer Distribution Based on Monetary Value')
plt.xlabel('Customer Segment')
plt.ylabel('Number of Customers')
plt.xticks(rotation=15)

for container in ax.containers:
    ax.bar_label(container, fmt='{:,.0f}', padding=3)

plt.tight_layout()
plt.show()
```

**Insight:**

Berdasarkan analisis segmentasi pelanggan menggunakan nilai monetary, sebanyak 93.358 pelanggan berhasil dikelompokkan menjadi tiga segmen, yaitu Low Value Customer sebanyak 30.812 pelanggan (33,00%), Medium Value Customer sebanyak 31.982 pelanggan (34,26%), dan High Value Customer sebanyak 30.564 pelanggan (32,74%).

Medium Value Customer memiliki jumlah pelanggan terbanyak, sedangkan High Value Customer memiliki jumlah pelanggan paling sedikit. Distribusi ketiga segmen relatif seimbang karena pengelompokan dilakukan menggunakan persentil 33% dan 67%. ### 2. Geospatial Analysis

Analisis geospasial dilakukan untuk mengetahui persebaran pelanggan berdasarkan lokasi geografis. Analisis ini menggunakan informasi kota, negara bagian, kode pos, serta koordinat geografis pelanggan. Hasil analisis diharapkan dapat memberikan gambaran mengenai wilayah dengan konsentrasi pelanggan tertinggi sehingga dapat menjadi bahan pertimbangan dalam menentukan strategi pemasaran dan distribusi.

```python
# Mengelompokkan koordinat berdasarkan kode pos
geolocation_grouped = (
    geolocation_clean
    .groupby('geolocation_zip_code_prefix')
    .agg(
        latitude=('geolocation_lat', 'mean'),
        longitude=('geolocation_lng', 'mean')
    )
    .reset_index()
)

# Menggabungkan data pelanggan dengan koordinat geografis
customer_geo_df = customers_clean.merge(
    geolocation_grouped,
    left_on='customer_zip_code_prefix',
    right_on='geolocation_zip_code_prefix',
    how='left'
)

# Menampilkan hasil penggabungan
customer_geo_df[
    ['customer_unique_id', 'customer_city', 'customer_state',
     'latitude', 'longitude']
].head()
```

```python
print("Jumlah data pelanggan:", len(customer_geo_df))
print("Koordinat latitude kosong:", customer_geo_df['latitude'].isna().sum())
print("Koordinat longitude kosong:", customer_geo_df['longitude'].isna().sum())
```

```python
# Menghapus pelanggan yang tidak memiliki koordinat
customer_geo_valid = customer_geo_df.dropna(
    subset=['latitude', 'longitude']
).copy()

# Menghitung jumlah pelanggan berdasarkan negara bagian
state_customer_summary = (
    customer_geo_valid
    .groupby('customer_state')['customer_unique_id']
    .nunique()
    .reset_index()
    .sort_values(by='customer_unique_id', ascending=False)
)

state_customer_summary.columns = [
    'customer_state',
    'total_customers'
]

# Menghitung persentase pelanggan
state_customer_summary['percentage'] = (
    state_customer_summary['total_customers'] /
    state_customer_summary['total_customers'].sum() * 100
)

state_customer_summary
```

```python
# Menghitung jumlah pelanggan dan rata-rata koordinat berdasarkan negara bagian
state_geo_summary = (
    customer_geo_valid
    .groupby('customer_state')
    .agg(
        total_customers=('customer_unique_id', 'nunique'),
        latitude=('latitude', 'mean'),
        longitude=('longitude', 'mean')
    )
    .reset_index()
)

state_geo_summary.head()
```

## Export Dataset for Streamlit Dashboard

Tahap ini dilakukan untuk menyimpan hasil pengolahan dan analisis data ke dalam format CSV. File yang dihasilkan akan digunakan sebagai sumber data dalam pengembangan dashboard menggunakan Streamlit.

```python
import os
import shutil

# Membuat folder untuk menyimpan hasil pengolahan
os.makedirs('dashboard_data', exist_ok=True)

# Dataset dan hasil analisis yang akan diekspor
export_data = {
    'sales_data.csv': sales_df,
    'monthly_sales.csv': monthly_sales,
    'category_sales.csv': category_sales,
    'rfm_data.csv': rfm_df,
    'segment_summary.csv': segment_summary,
    'customer_value_summary.csv': customer_value_summary,
    'delivery_summary.csv': delivery_summary,
    'state_customer_summary.csv': state_customer_summary,
    'state_geo_summary.csv': state_geo_summary
}

# Menyimpan setiap DataFrame ke CSV
for filename, dataframe in export_data.items():
    dataframe.to_csv(
        f'dashboard_data/{filename}',
        index=False
    )

# Membuat ZIP
shutil.make_archive(
    'dashboard_data',
    'zip',
    'dashboard_data'
)

print("Seluruh dataset berhasil diekspor!")
```
